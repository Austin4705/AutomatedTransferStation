```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import cpbd 
```

```python
def calculate_cpbd(image):
    if len(image.shape) == 3:
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    else:
        gray = image
    return cpbd.compute(gray)
```

```python
image_dir = Path("Dataset/GR08-6/collection_2026-03-02_11-30-59")

image_name='image_2175_2026-03-02_12-39-02.png'
image_path = image_dir / image_name
image = cv2.imread(str(image_path))
cpbd_value = calculate_cpbd(image)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title(f"CPBD value for {image_name}: {cpbd_value}")
plt.axis('off')
plt.show()
```
