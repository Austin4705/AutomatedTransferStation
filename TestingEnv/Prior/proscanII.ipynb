```python
from serial import Serial
import threading
from queue import Queue
import json
import os         
import time
from IPython import display


device = Serial(port='COM3', baudrate=9600, timeout=.1) 

def callback(data):
    display.clear_output()
    print(data)

def send_to_station(command):
    device.write(bytes(f"{command}\r\n", 'ascii'))

def listen_serial():
    while True:
        reading = device.readline()
        print(f'Data: {reading}')

thread = threading.Thread(target=listen_serial)
thread.daemon = True
thread.start()

time.sleep(1)
print("Done")
send_to_station("MACRO")
```
