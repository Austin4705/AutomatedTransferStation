```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
```

```python

def plot(points):
    # Separate x and y for plotting
    x_vals = [pt[1] for pt in points]
    y_vals = [pt[0] for pt in points]
    plt.figure(figsize=(6,4))
    plt.plot(x_vals, y_vals, marker='o', linestyle='-')
    plt.xlabel('X')
    plt.ylabel('Y')

    # Define Gaussian function
    def gaussian(x, amp, mean, std):
        return amp * np.exp(-((x - mean) ** 2) / (2 * std ** 2))

    # Convert to numpy arrays for fitting
    x_vals_np = np.array(x_vals)
    y_vals_np = np.array(y_vals)

    # Initial guess for parameters: amplitude, mean, std
    init_guess = [y_vals_np.max(), x_vals_np[np.argmax(y_vals_np)], 0.05]

    # Fit the Gaussian
    try:
        popt, pcov = curve_fit(gaussian, x_vals_np, y_vals_np, p0=init_guess)
        fit_success = True
    except Exception as e:
        print("Gaussian fit error:", e)
        fit_success = False

    # Plot Gaussian fit if successful
    if fit_success:
        x_fit = np.linspace(min(x_vals_np), max(x_vals_np), 200)
        y_fit = gaussian(x_fit, *popt)
        plt.plot(x_fit, y_fit, color='red', label='Gaussian fit')
        plt.legend()
        print(f"Fitted parameters: amplitude={popt[0]:.2f}, mean={popt[1]:.5f}, std={popt[2]:.5f}")
    plt.title('Graph of XY Points')
    plt.grid(True)
    plt.show()
```
